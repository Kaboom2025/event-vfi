# 00b — Tar Processed Data
Creates `processed.tar` on Drive so training sessions can copy one file instead of thousands.

**Run once on a CPU runtime** — no GPU needed.

```python
from google.colab import drive
drive.mount('/content/drive')

import subprocess, os

processed_dir = '/content/drive/MyDrive/493Project/data/vimeo/processed'
tar_on_drive  = '/content/drive/MyDrive/493Project/data/vimeo/processed.tar'

if os.path.exists(tar_on_drive):
    size_gb = os.path.getsize(tar_on_drive) / 1e9
    print(f'processed.tar already exists on Drive ({size_gb:.1f} GB) — nothing to do.')
else:
    subprocess.run(['apt-get', 'install', '-qq', '-y', 'pv'], capture_output=True)

    print('Calculating total size...')
    result = subprocess.run(['du', '-sb', processed_dir], capture_output=True, text=True)
    total_bytes = int(result.stdout.split()[0])
    print(f'Total: {total_bytes / 1e9:.1f} GB')

    print('Creating tar on Drive (one-time cost)...')
    subprocess.run(
        f'tar cf - -C "{processed_dir}" . | pv -f -s {total_bytes} > "{tar_on_drive}"',
        shell=True, check=True,
    )
    size_gb = os.path.getsize(tar_on_drive) / 1e9
    print(f'Done. Saved {tar_on_drive} ({size_gb:.1f} GB)')
```
